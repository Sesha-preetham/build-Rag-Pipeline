````python
import json
import re
import time
import pandas as pd
from openai import OpenAI

client = OpenAI()
````

````python
def extract_json(text):
    """
    Extract JSON from an LLM response.
    Handles plain JSON and JSON inside markdown fences.
    """
    if not text:
        return None

    text = text.strip()

    # Remove markdown fences
    text = re.sub(r"^```json\s*", "", text)
    text = re.sub(r"^```\s*", "", text)
    text = re.sub(r"\s*```$", "", text)

    # Find JSON object
    match = re.search(r"\{.*\}", text, re.DOTALL)

    if match:
        try:
            return json.loads(match.group())
        except json.JSONDecodeError:
            return None

    return None
````

````python
def evaluate_with_llm_judge(
    question,
    expected_answer,
    rag_answer,
    retrieved_sources,
    expected_sources,
    judge_model="gpt-4o-mini"
):
    """
    Evaluate RAG answer using LLM-as-Judge.

    Returns:
        answer_correctness: 1-5
        groundedness: 1-5
        citation_correctness: 1-5
        overall_score: 1-5
        explanation: text
    """

    prompt = f"""
You are an expert evaluator for a Retrieval-Augmented Generation (RAG) system.

Evaluate the RAG response against the golden reference.

QUESTION:
{question}

GOLDEN / EXPECTED ANSWER:
{expected_answer}

EXPECTED SOURCE DOCUMENTS:
{expected_sources}

RAG ANSWER:
{rag_answer}

SOURCE DOCUMENTS RETRIEVED BY RAG:
{retrieved_sources}

Evaluate using the following rules.

1. ANSWER_CORRECTNESS (1-5)
5 = Fully correct and contains the important facts from the expected answer.
4 = Mostly correct, with only a small omission or minor wording difference.
3 = Partially correct; important information is missing.
2 = Mostly incorrect.
1 = Incorrect, irrelevant, or hallucinated.

2. GROUNDEDNESS (1-5)
5 = Answer is clearly supported by the retrieved sources and does not introduce unsupported facts.
4 = Mostly grounded with a small unsupported inference.
3 = Mix of supported and unsupported claims.
2 = Mostly unsupported.
1 = Clearly hallucinated or unsupported.

IMPORTANT:
Groundedness should depend on whether the answer can reasonably be supported
by the retrieved documents, not just whether the answer happens to be correct.

3. CITATION_CORRECTNESS (1-5)
5 = All important expected source documents were retrieved.
4 = Most required sources were retrieved.
3 = Some required sources were retrieved.
2 = Retrieved sources are mostly wrong or irrelevant.
1 = No relevant expected source was retrieved.

4. OVERALL_SCORE (1-5)
Overall quality considering correctness, groundedness and retrieval quality.

IMPORTANT FOR MULTI-CURRENCY QUESTIONS:
If the golden answer says currencies must NOT be directly subtracted without an
exchange rate, the RAG answer must recognize that. Penalize an answer that
performs an invalid direct subtraction across currencies.

Return ONLY valid JSON in this exact format:

{{
    "answer_correctness": 1,
    "groundedness": 1,
    "citation_correctness": 1,
    "overall_score": 1,
    "explanation": "short explanation"
}}
"""

    try:
        response = client.chat.completions.create(
            model=judge_model,
            messages=[
                {
                    "role": "system",
                    "content": "You are a strict and unbiased RAG evaluation judge."
                },
                {
                    "role": "user",
                    "content": prompt
                }
            ],
            temperature=0,
            response_format={"type": "json_object"}
        )

        result = json.loads(response.choices[0].message.content)

        return result

    except Exception as e:
        print(f"Judge error: {e}")

        return {
            "answer_correctness": None,
            "groundedness": None,
            "citation_correctness": None,
            "overall_score": None,
            "explanation": f"Judge failed: {str(e)}"
        }
````

````python
def calculate_retrieval_metrics(expected_sources, retrieved_sources):
    """
    Calculate deterministic retrieval KPIs.
    """

    expected_set = set(expected_sources)
    retrieved_set = set(retrieved_sources)

    matched_sources = expected_set.intersection(retrieved_set)

    # At least one expected document retrieved
    retrieval_hit = len(matched_sources) > 0

    # Percentage of required documents retrieved
    source_recall = (
        len(matched_sources) / len(expected_set)
        if expected_set else 0
    )

    # Precision of retrieved source documents
    source_precision = (
        len(matched_sources) / len(retrieved_set)
        if retrieved_set else 0
    )

    # Did we retrieve every expected source?
    all_expected_sources_retrieved = expected_set.issubset(retrieved_set)

    return {
        "retrieval_hit": retrieval_hit,
        "source_recall": source_recall,
        "source_precision": source_precision,
        "all_expected_sources_retrieved": all_expected_sources_retrieved,
        "matched_sources": list(matched_sources)
    }
````

````python
def run_golden_set_evaluation(
    golden_set,
    index,
    k=5,
    judge_model="gpt-4o-mini"
):
    results = []

    total_questions = len(golden_set)

    print(f"Running evaluation for {total_questions} questions...\n")

    for i, item in enumerate(golden_set, 1):

        question_id = item["id"]
        question = item["question"]

        print("=" * 100)
        print(
            f"[{i}/{total_questions}] "
            f"{question_id} | {item['difficulty'].upper()}"
        )
        print(f"Question: {question}")

        # -------------------------------------------------------
        # 1. Ask RAG
        # -------------------------------------------------------
        start_time = time.time()

        rag_response = ask_rag(
            question,
            index,
            k=k
        )

        latency = time.time() - start_time

        rag_answer = rag_response.get("answer", "")

        # Sources returned by your ask_rag implementation
        retrieved_sources = rag_response.get("sources", [])

        # Remove duplicates while preserving order
        retrieved_sources = list(dict.fromkeys(retrieved_sources))

        # -------------------------------------------------------
        # 2. Calculate deterministic retrieval metrics
        # -------------------------------------------------------
        retrieval_metrics = calculate_retrieval_metrics(
            item["expected_sources"],
            retrieved_sources
        )

        # -------------------------------------------------------
        # 3. LLM as Judge
        # -------------------------------------------------------
        judge_result = evaluate_with_llm_judge(
            question=question,
            expected_answer=item["expected_answer"],
            rag_answer=rag_answer,
            retrieved_sources=retrieved_sources,
            expected_sources=item["expected_sources"],
            judge_model=judge_model
        )

        # -------------------------------------------------------
        # 4. Cost
        # -------------------------------------------------------
        try:
            rag_cost = cost_usd(rag_response)
        except Exception:
            rag_cost = None

        # -------------------------------------------------------
        # 5. Save result
        # -------------------------------------------------------
        result = {
            "id": question_id,
            "difficulty": item["difficulty"],
            "question": question,
            "expected_answer": item["expected_answer"],
            "rag_answer": rag_answer,

            "expected_sources": ", ".join(item["expected_sources"]),
            "retrieved_sources": ", ".join(retrieved_sources),

            "retrieval_hit": retrieval_metrics["retrieval_hit"],
            "source_recall": retrieval_metrics["source_recall"],
            "source_precision": retrieval_metrics["source_precision"],
            "all_expected_sources_retrieved":
                retrieval_metrics["all_expected_sources_retrieved"],
            "matched_sources": ", ".join(
                retrieval_metrics["matched_sources"]
            ),

            "answer_correctness":
                judge_result.get("answer_correctness"),
            "groundedness":
                judge_result.get("groundedness"),
            "citation_correctness":
                judge_result.get("citation_correctness"),
            "overall_score":
                judge_result.get("overall_score"),
            "judge_explanation":
                judge_result.get("explanation"),

            "latency_s": latency,
            "cost_usd": rag_cost,

            "tokens_in": rag_response.get("tokens_in"),
            "tokens_out": rag_response.get("tokens_out")
        }

        results.append(result)

        # -------------------------------------------------------
        # Print result
        # -------------------------------------------------------
        print(f"\nRAG Answer:")
        print(rag_answer)

        print(f"\nExpected Sources:")
        print(item["expected_sources"])

        print(f"\nRetrieved Sources:")
        print(retrieved_sources)

        print(f"\nKPIs:")
        print(
            f"  Retrieval Hit: "
            f"{retrieval_metrics['retrieval_hit']}"
        )
        print(
            f"  Source Recall: "
            f"{retrieval_metrics['source_recall'] * 100:.1f}%"
        )
        print(
            f"  Answer Correctness: "
            f"{judge_result.get('answer_correctness')}/5"
        )
        print(
            f"  Groundedness: "
            f"{judge_result.get('groundedness')}/5"
        )
        print(
            f"  Citation Correctness: "
            f"{judge_result.get('citation_correctness')}/5"
        )
        print(
            f"  Overall Score: "
            f"{judge_result.get('overall_score')}/5"
        )
        print(f"  Latency: {latency:.2f}s")

    return pd.DataFrame(results)
````

````python
results_df = run_golden_set_evaluation(
    golden_set=GOLDEN_SET,
    index=index,
    k=5,
    judge_model="gpt-4o-mini"
)

results_df.head()
````

````python
overall_kpis = {
    "Total Questions": len(results_df),

    "Answer Correctness (%)":
        results_df["answer_correctness"].mean() / 5 * 100,

    "Groundedness (%)":
        results_df["groundedness"].mean() / 5 * 100,

    "Citation Correctness (%)":
        results_df["citation_correctness"].mean() / 5 * 100,

    "Overall LLM Judge Score (%)":
        results_df["overall_score"].mean() / 5 * 100,

    "Retrieval Hit Rate (%)":
        results_df["retrieval_hit"].mean() * 100,

    "Average Source Recall (%)":
        results_df["source_recall"].mean() * 100,

    "Average Source Precision (%)":
        results_df["source_precision"].mean() * 100,

    "All Required Sources Retrieved (%)":
        results_df["all_expected_sources_retrieved"].mean() * 100,

    "Average Latency (seconds)":
        results_df["latency_s"].mean(),

    "P50 Latency (seconds)":
        results_df["latency_s"].median(),

    "Total RAG Cost (USD)":
        results_df["cost_usd"].sum(),

    "Average Cost per Question (USD)":
        results_df["cost_usd"].mean()
}

overall_kpis_df = pd.DataFrame(
    list(overall_kpis.items()),
    columns=["KPI", "Value"]
)

overall_kpis_df
````

````python
difficulty_summary = (
    results_df
    .groupby("difficulty")
    .agg(
        questions=("id", "count"),

        answer_correctness=("answer_correctness", "mean"),
        groundedness=("groundedness", "mean"),
        citation_correctness=("citation_correctness", "mean"),
        overall_score=("overall_score", "mean"),

        retrieval_hit_rate=("retrieval_hit", "mean"),
        source_recall=("source_recall", "mean"),

        avg_latency_s=("latency_s", "mean"),
        total_cost_usd=("cost_usd", "sum")
    )
    .round(4)
)

# Convert scores and rates to percentages
difficulty_summary["answer_correctness"] = (
    difficulty_summary["answer_correctness"] / 5 * 100
)

difficulty_summary["groundedness"] = (
    difficulty_summary["groundedness"] / 5 * 100
)

difficulty_summary["citation_correctness"] = (
    difficulty_summary["citation_correctness"] / 5 * 100
)

difficulty_summary["overall_score"] = (
    difficulty_summary["overall_score"] / 5 * 100
)

difficulty_summary["retrieval_hit_rate"] = (
    difficulty_summary["retrieval_hit_rate"] * 100
)

difficulty_summary["source_recall"] = (
    difficulty_summary["source_recall"] * 100
)

difficulty_summary
````
